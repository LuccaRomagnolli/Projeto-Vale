# 07 - Governança do Modelo e Critério de Promoção

Notebook para auditoria: regra de promoção, estabilidade temporal e rastreabilidade.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().resolve().parents[0]
policy = ROOT / 'docs' / 'politica_promocao_modelo.md'
benchmark = json.loads((ROOT / 'reports' / 'model_benchmark_report.json').read_text())
tuning = json.loads((ROOT / 'reports' / 'hist_gbdt_tuning_report.json').read_text())
backtest = pd.read_csv(ROOT / 'reports' / 'hist_gbdt_backtest_report.csv')

print('Política existe?', policy.exists())
print('Campeão benchmark:', benchmark['winner']['model_name'])
print('Melhor tuning:', tuning['best_candidate']['candidate'])
```

```python
# Gate de estabilidade temporal (cálculo manual para apresentação)
recall_std = float(backtest['test_recall'].std(ddof=0))
precision_std = float(backtest['test_precision'].std(ddof=0))
gate = {
    'recall_std': recall_std,
    'precision_std': precision_std,
    'limit_recall_std': 0.03,
    'limit_precision_std': 0.05,
    'passed': recall_std <= 0.03 and precision_std <= 0.05,
}
gate
```
