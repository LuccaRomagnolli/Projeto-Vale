# Extracao e visualizacao inicial dos dados

Notebook simples para conferir os arquivos extraidos e visualizar amostras com `head()`.

```python
from pathlib import Path
import pandas as pd

pd.set_option('display.max_columns', 80)
pd.set_option('display.width', 140)
```

```python
PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
RAW_BASE = PROJECT_ROOT / 'data' / 'raw' / 'datasets' / 'datasets'

apontamentos_dir = RAW_BASE / 'apontamentos'
telemetria_dir = RAW_BASE / 'telemetria'

print('PROJECT_ROOT:', PROJECT_ROOT)
print('RAW_BASE:', RAW_BASE)
print('Arquivos em apontamentos:')
for p in sorted(apontamentos_dir.glob('*')):
    print(' -', p.name)

print('\nArquivos em telemetria:')
for p in sorted(telemetria_dir.glob('*')):
    print(' -', p.name)
```

```python
# Apontamentos: tenta parquet e faz fallback para xlsx
apont_parquet = apontamentos_dir / 'desenvolver_apontamentos.parquet'
apont_xlsx = apontamentos_dir / 'desenvolver_apontamentos.xlsx'

try:
    df_apont = pd.read_parquet(apont_parquet)
    print('Fonte usada: parquet')
except Exception as e:
    print(f'Parquet indisponivel ({e}). Usando XLSX...')
    df_apont = pd.read_excel(apont_xlsx)

print('Shape apontamentos:', df_apont.shape)
df_apont.head()
```

```python
# Regras/alertas
dontgo_xlsx = telemetria_dir / 'desenvolver_dontgo.xlsx'
df_dontgo = pd.read_excel(dontgo_xlsx)

print('Shape dontgo:', df_dontgo.shape)
df_dontgo.head()
```

```python
# Telemetria: carrega um arquivo mensal e mostra head
telemetry_file = telemetria_dir / 'telemetry_jan.parquet'
df_telemetry = pd.read_parquet(telemetry_file)

print('Arquivo:', telemetry_file.name)
print('Shape telemetry_jan:', df_telemetry.shape)
df_telemetry.head()
```
